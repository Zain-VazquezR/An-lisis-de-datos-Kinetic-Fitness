# Simulación de datos

## Librerías

```python
import numpy as np
from datetime import datetime
import calendar
from datetime import datetime, timedelta
```

## Generación de datos

```python
fechas = []
costos = []
mensualidades = []

meses_descuento = [1, 4, 7]
dias_alta_densidad = [1, 2, 3, 14, 15, 16, 28, 30, 31]
pesos_meses = np.array([
    2.3,  # Enero
    1.4,  # Febrero
    1.1,  # Marzo
    2.0,  # Abril
    1.1,  # Mayo
    1.0,  # Junio
    2.3,  # Julio
    1.5,  # Agosto
    1.1,  # Septiembre
    0.9,  # Octubre
    1.0,  # Noviembre
    1.2   # Diciembre
])

probabilidades_meses = pesos_meses / pesos_meses.sum()
for _ in range(2500):
    mes = np.random.choice(
        np.arange(1, 13),
        p=probabilidades_meses
    )
    ultimo_dia = calendar.monthrange(2025, mes)[1]

    dias_validos = list(range(1, ultimo_dia + 1))
    dias_alta = [
        dia for dia in dias_alta_densidad
        if dia <= ultimo_dia
    ]

    dias_normales = [
        dia for dia in dias_validos
        if dia not in dias_alta
    ]

    if np.random.random() < 0.60:
        dia = np.random.choice(dias_alta)
    else:
        dia = np.random.choice(dias_normales)

    fecha = datetime(2025, mes, dia)
    fechas.append(fecha)

fechas.sort()

precios = [499, 1099, 1999, 3199]
probabilidades = [0.55, 0.23, 0.13, 0.09]
for fecha in fechas:

    costo = np.random.choice(
        precios,
        p=probabilidades
    )

    costos.append(costo)

    if costo == 499:
        mensualidades.append(1)
    elif costo == 1099:
        mensualidades.append(3)
    elif costo == 1999:
        mensualidades.append(6)
    elif costo == 3199:
        mensualidades.append(12)

print("Fecha\tCosto\tMensualidad")

for i in range(len(fechas)):
    print(
        f"{fechas[i].strftime('%d/%m/%Y')}\t"
        f"{costos[i]}\t"
        f"{mensualidades[i]}"
    )
```

```python
sedes = []

for _ in range(len(fechas)):
    sedes.append(np.random.randint(1, 4))

print("Fecha\tCosto\tMensualidad\tSede")

for i in range(len(fechas)):
    print(
        f"{fechas[i].strftime('%d/%m/%Y')}\t"
        f"{costos[i]}\t"
        f"{mensualidades[i]}\t"
        f"{sedes[i]}"
    )
```

```python
meses = []

for fecha in fechas:
    meses.append(fecha.month)

nombres_meses = [
    "Enero", "Febrero", "Marzo", "Abril",
    "Mayo", "Junio", "Julio", "Agosto",
    "Septiembre", "Octubre", "Noviembre", "Diciembre"
]

print("Fecha\tCosto\tMensualidad\tSede\tMes")

for i in range(len(fechas)):
    nombre_mes = nombres_meses[meses[i] - 1]

    print(
        f"{fechas[i].strftime('%d/%m/%Y')}\t"
        f"{costos[i]}\t"
        f"{mensualidades[i]}\t"
        f"{sedes[i]}\t"
        f"{nombre_mes}"
    )
```

```python
descuentos = []

for mes in meses:
    if mes == 1 or mes == 4 or mes == 7:
        descuentos.append(0.9)
    else:
        descuentos.append(1)

print("Fecha\tCosto_productos\tMensualidad\tSede\tMes\tDescuento")

for i in range(len(fechas)):
    nombre_mes = nombres_meses[meses[i] - 1]

    print(
        f"{fechas[i].strftime('%d/%m/%Y')}\t"
        f"{costos[i]}\t"
        f"{mensualidades[i]}\t"
        f"{sedes[i]}\t"
        f"{nombre_mes}\t"
        f"{descuentos[i]}"
    )
```

```python
n_ventas = len(fechas)

usarios_antiguos = [
    f"G{i:06d}" for i in range(1, 305)
]

usarios_nuevos = [
    f"G{i:06d}" for i in range(305, 1200)
]

usarios = usarios_antiguos + usarios_nuevos

ultima_compra = {}

ultima_mensualidad = {}

numero_compras = {}


ids_usuario = []

for i in range(n_ventas):

    fecha = fechas[i]
    mensualidad_actual = mensualidades[i]

    usuarios_disponibles = []

    for usuario in ultima_compra:

        fecha_anterior = ultima_compra[usuario]
        membresia_anterior = ultima_mensualidad[usuario]

        dias_transcurridos = (fecha - fecha_anterior).days

        if membresia_anterior == 1:

            dias_minimos = 28
            dias_maximos = 32

        elif membresia_anterior == 3:

            dias_minimos = 89
            dias_maximos = 93

        elif membresia_anterior == 6:

            dias_minimos = 180
            dias_maximos = 184

        elif membresia_anterior == 12:

            dias_minimos = 366
            dias_maximos = 366

        if dias_minimos <= dias_transcurridos <= dias_maximos:
            usuarios_disponibles.append(usuario)


    probabilidad = np.random.random()
    if usuarios_disponibles and probabilidad < 0.55:

        usuario = np.random.choice(usuarios_disponibles)

    elif ultima_compra and probabilidad < 0.70:

        usuarios_reaparecen = []

        for usuario_existente in ultima_compra:

            fecha_anterior = ultima_compra[usuario_existente]
            membresia_anterior = ultima_mensualidad[usuario_existente]

            dias_transcurridos = (fecha - fecha_anterior).days

            if membresia_anterior == 1:
                dias_minimos_reaparicion = 33

            elif membresia_anterior == 3:
                dias_minimos_reaparicion = 94

            elif membresia_anterior == 6:
                dias_minimos_reaparicion = 185

            else:
                dias_minimos_reaparicion = 367

            if dias_transcurridos >= dias_minimos_reaparicion:
                usuarios_reaparecen.append(usuario_existente)

        if usuarios_reaparecen:
            usuario = np.random.choice(usuarios_reaparecen)
        else:
            usuario = None

    else:
        usuario = None



    if usuario is None:

        if np.random.random() < 0.85:

            disponibles = [
                u for u in usarios_nuevos
                if u not in ultima_compra
            ]

        else:

            disponibles = [
                u for u in usarios_antiguos
                if u not in ultima_compra
            ]

        if not disponibles:

            disponibles = usuarios

        usuario = np.random.choice(disponibles)



    ids_usuario.append(usuario)

    ultima_compra[usuario] = fecha

    ultima_mensualidad[usuario] = mensualidad_actual

    if usuario not in numero_compras:

        numero_compras[usuario] = 1

    else:

        numero_compras[usuario] += 1

print(
    "Fecha\tCosto\tMensualidad\tSede\tDescuento\tID_Usuario"
)

for i in range(n_ventas):

    print(
        f"{fechas[i].strftime('%d/%m/%Y')}\t"
        f"{costos[i]}\t"
        f"{mensualidades[i]}\t"
        f"{sedes[i]}\t"
        f"{descuentos[i]}\t"
        f"{ids_usuario[i]}"
    )
```

### Tabla de datos

```python
columnas = {
    "fecha": fechas,
    "costo": costos,
    "sede": sedes,
    "mensualidad": mensualidades,
    "descuento": descuentos,
    "mes": meses,
    "id": ids_usuario,
    "id_usuario": ids_usuario
}


entrada = input(
    "Escribe las columnas que deseas consultar "
    "(separadas por coma) o escribe 'completa':\n"
).lower().strip()


if entrada == "completa":

    seleccionadas = [
        "fecha",
        "costo",
        "sede",
        "mensualidad",
        "descuento",
        "mes",
        "id_usuario"
    ]

else:

    seleccionadas = [
        columna.strip()
        for columna in entrada.split(",")
    ]



columnas_invalidas = [
    columna
    for columna in seleccionadas
    if columna not in columnas
]

if columnas_invalidas:

    print("\nColumnas no reconocidas:")

    for columna in columnas_invalidas:
        print("-", columna)

else:


    encabezados = []

    for columna in seleccionadas:

        if columna == "id":
            encabezados.append("ID_Usuario")

        elif columna == "id_usuario":
            encabezados.append("ID_Usuario")

        elif columna == "fecha":
            encabezados.append("Fecha")

        elif columna == "costo":
            encabezados.append("Costo")

        elif columna == "sede":
            encabezados.append("Sede")

        elif columna == "mensualidad":
            encabezados.append("Mensualidad")

        elif columna == "descuento":
            encabezados.append("Descuento")

        elif columna == "mes":
            encabezados.append("Mes")



    print("\n" + "\t".join(encabezados))



    for i in range(len(fechas)):

        fila = []

        for columna in seleccionadas:

            valor = columnas[columna][i]
            if columna == "fecha":

                valor = valor.strftime("%d/%m/%Y")

            fila.append(str(valor))

        print("\t".join(fila))
```

# Consulta de datos y revisión

## Librerías

```python
from collections import Counter
from IPython.display import display
from ipywidgets import widgets
```

## Consulta

```python

conteo_ids = Counter(ids_usuario)

ids_que_se_repiten = {user_id: count for user_id, count in conteo_ids.items() if count > 1}

print("Todos los IDs de usuario que se repiten son:")
if ids_que_se_repiten:
    for user_id, count in ids_que_se_repiten.items():
        print(f"ID: {user_id}, Repeticiones: {count}")
else:
    print("No hay IDs de usuario que se repitan.")
```

### Buscador de usuarios

```python

input_id = widgets.Text(
    value='',
    placeholder='Introduce el ID de usuario (ej. G000416)',
    description='ID de Usuario:',
    disabled=False
)

search_button = widgets.Button(description="Buscar")

def buscar_id(sender):
    user_id_to_find = input_id.value.strip()

    if not user_id_to_find:
        print("Por favor, introduce un ID de usuario.")
        return

    found_entries = []
    for i in range(len(ids_usuario)):
        if ids_usuario[i] == user_id_to_find:
            found_entries.append({
                "fecha": fechas[i],
                "costo": costos[i],
                "mensualidad": mensualidades[i],
                "sede": sedes[i],
                "descuento": descuentos[i],
                "mes": nombres_meses[meses[i] - 1],
                "id_usuario": ids_usuario[i]
            })

    if found_entries:
        print(f"\nInformación para el ID de usuario: {user_id_to_find}\n")
        print("Fecha\tCosto\tMensualidad\tSede\tDescuento\tMes\tID_Usuario")
        for entry in found_entries:
            print(
                f"{entry['fecha'].strftime('%d/%m/%Y')}\t"
                f"{entry['costo']}\t"
                f"{entry['mensualidad']}\t"
                f"{entry['sede']}\t"
                f"{entry['descuento']}\t"
                f"{entry['mes']}\t"
                f"{entry['id_usuario']}"
            )
    else:
        print(f"No se encontró información para el ID de usuario: {user_id_to_find}")

search_button.on_click(buscar_id)

display(input_id, search_button)
```
